```python
user_question = """
We are solving a forward problem for the Korteweg-De Vries (KdV) equation:

$$
u_t + 6uu_x + u_{xxx} = 0
$$

**Domain:**
- Spatial: $x \in [0, 20]$
- Temporal: $t \in [0, 5]$

**Boundary Conditions:**

$$
u(0,x) = g_0(x) \quad \text{(Initial Condition)}
$$

$$
u(t,0) = g_1(t) \quad \text{(BC at } x=0\text{)}
$$

$$
u(t,20) = g_2(t) \quad \text{(BC at } x=20\text{)}
$$

$$
u_x(t,20) = g_3(t) \quad \text{(Neumann BC at } x=20\text{)}
$$

**Analytical Solution:**

The solution is given by a two-soliton interaction formula:

$$
u(x,t) = \frac{2(c_1 - c_2) \left[ c_1 \cosh^2\left(\frac{\sqrt{c_2}\zeta_2}{2}\right) + c_2 \sinh^2\left(\frac{\sqrt{c_1}\zeta_1}{2}\right) \right]}{\left[ \left(\sqrt{c_1} - \sqrt{c_2}\right) \cosh\left(\frac{\sqrt{c_1}\zeta_1 + \sqrt{c_2}\zeta_2}{2}\right) + \left(\sqrt{c_1} + \sqrt{c_2}\right) \cosh\left(\frac{\sqrt{c_1}\zeta_1 - \sqrt{c_2}\zeta_2}{2}\right) \right]^2}
$$


where:
- $\zeta_1 = x - c_1 t - x_1$
- $\zeta_2 = x - c_2 t - x_2$

**Parameters:**
- $c_1 = 6.0$ (velocity of first soliton)
- $c_2 = 2.0$ (velocity of second soliton)
- $x_1 = -2.0$ (initial position of first soliton)
- $x_2 = 2.0$ (initial position of second soliton)

The boundary conditions $g_0, g_1, g_2, g_3$ are derived by evaluating the analytical solution at the appropriate boundaries.

MANDATORY: Make sure to report the loss and errors every 500 iterations and you cannonly use optimizers that work natively in JAX (NO SciPY)
CONSTRAINTS: If using latin hypercube sampling dont use any criterion such as 'maximin'. 
"""
```
